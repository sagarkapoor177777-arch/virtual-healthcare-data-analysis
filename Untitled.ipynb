```python
df['readmitted'].value_counts()
```

```python
redmission_pct = df['readmitted'].value_counts(normalize= True)*100
redmission_pct
```

```python
df['readmitted'].value_counts().plot(kind='bar')
plt.title('Patient Readmission Distribution')
plt.xlabel('Readmission Status')
plt.ylabel('Number of Patients')
plt.xticks(rotation= 0 )
plt.show()
plt.close()
```

```python
age_readmission = pd.crosstab(df['age'],df['readmitted'],normalize='index') *100 
age_readmission
```

```python
age_readmission.plot(kind='bar',figsize= (12,6))
plt.title('Readmission Rate by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Percentage of Patients')
plt.xticks(rotation= 45)
plt.legend(title='Readmission')
plt.tight_layout()
plt.show()
```

```python
med_readmission = pd.crosstab(
    df['diabetesMed'],
    df['readmitted'], 
    normalize= 'index'
) * 100
med_readmission
```

```python
med_readmission.plot(kind='bar', figsize=(10,6))

plt.title('Readmission Rate by Diabetes Medication')
plt.xlabel('Diabetes Medication')
plt.ylabel('Percentage of patients')
plt.xticks(rotation=0)
plt.legend(title='Readmission')
plt.tight_layout()
plt.show()
```

```python
stay_readmission = pd.crosstab(
    df['time_in_hospital'],
    df['readmitted'], 
    normalize='index'
)*100
stay_readmission
```

```python
stay_readmission.plot(kind='bar',figsize=(12,6))
plt.title('Readmission Rate by Length of Hospital Stay')
plt.xlabel('Days in Hospital')
plt.ylabel('Percentage of patients')
plt.xticks(rotation= 0 )
plt.legend(title='Readmission')
plt.tight_layout()
plt.show()
```

```python
inpatient_readmission = pd.crosstab(
    df['number_inpatient'], 
    df['readmitted'], 
    normalize = 'index'
)* 100
inpatient_readmission
```

```python
inpatient_readmission.plot(kind= 'bar',figsize=(12,6))
plt.title('Readmission Rate by Previous Inpatient Visiits')
plt.xlabel('Number of Previous Inpatient Visits')
plt.ylabel('Percetage of Patients')
plt.legend(title='Readmission')
plt.xticks(rotation= 0)
plt.tight_layout()
plt.show()
```

```python
emergency_readmission = pd.crosstab(
    df['number_emergency'], 
    df['readmitted'], 
    normalize='index'
) *100
emergency_readmission
```

```python
emergency_readmission.plot(kind='bar',figsize= (12,6))
plt.title('Readmission Rate by Previous Emergency Visits')
plt.xlabel('Number of Previous Emergency Visits')
plt.ylabel('Percentage of Patients')
plt.legend(title = 'Readmission')
plt.xticks(rotation= 0)
plt.tight_layout()
plt.show()
```

```python
diagnosis_readmission  = pd.crosstab(
    df['number_diagnoses'], 
    df['readmitted'], 
    normalize='index'
) * 100 
diagnosis_readmission
```

```python
diagnosis_readmission.plot(kind='bar',figsize=(12,6))
plt.title('Readmission Rate by Number of Diagnoses')
plt.xlabel('Number of Diagnoues')
plt.ylabel('Percentage of Patients')
plt.legend(title= 'Readmission')
plt.xticks(rotation= 0)
plt.tight_layout()
plt.show()
```
